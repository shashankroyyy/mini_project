```python
import pandas as pd

# Load the dataset
df = pd.read_csv("hyd_metro_crowd_dataset.csv")

# Display first 10 rows
df.head(10)
```

```python
from sklearn.preprocessing import LabelEncoder

le_station = LabelEncoder()
le_day = LabelEncoder()
le_crowd = LabelEncoder()

df["station_encoded"] = le_station.fit_transform(df["station"])
df["day_type_encoded"] = le_day.fit_transform(df["day_type"])
df["crowd_encoded"] = le_crowd.fit_transform(df["crowd_level"])

df.head(10)
```

```python
X = df[["station_encoded", "hour", "day_type_encoded"]]
Y = df["crowd_encoded"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.3, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_train, Y_train)
```

```python
from sklearn.metrics import accuracy_score

predictions = model.predict(X_test)
accuracy = accuracy_score(Y_test, predictions)

print("Model Accuracy:", accuracy)
```

```python
# ============================
# STEP 3.1: DEFINE METRO GRAPH
# ============================

# Graph represented as: station -> list of (neighbor_station, travel_time_minutes)
metro_graph = {
    "Miyapur": [("JNTU College", 3)],
    "JNTU College": [("Miyapur", 3), ("KPHB Colony", 3)],
    "KPHB Colony": [("JNTU College", 3), ("SR Nagar", 3)],
    "SR Nagar": [("KPHB Colony", 3), ("Ameerpet", 3)],
    "Ameerpet": [("SR Nagar", 3), ("LB Nagar", 5), ("MGBS", 5), ("Raidurg", 6)],
    "LB Nagar": [("Ameerpet", 5)],

    "Nagole": [("Uppal", 3)],
    "Uppal": [("Nagole", 3), ("Stadium", 3)],
    "Stadium": [("Uppal", 3), ("MGBS", 4)],
    "MGBS": [("Stadium", 4), ("Ameerpet", 5), ("JBS Parade Ground", 4)],

    "Raidurg": [("Ameerpet", 6)],

    "JBS Parade Ground": [("MGBS", 4)],
}
```

```python
import heapq

def dijkstra_shortest_path(graph, source, destination):
    """
    Returns (total_time, path_list) using Dijkstra's algorithm.
    graph: dict of station -> list of (neighbor, weight)
    source, destination: station names (strings)
    """

    # Min-heap priority queue: (current_time, current_station, path_taken)
    heap = [(0, source, [source])]

    # Track the best known time to each station
    visited = {}

    while heap:
        current_time, station, path = heapq.heappop(heap)

        # If we already found a faster way before, skip this one
        if station in visited and visited[station] <= current_time:
            continue

        visited[station] = current_time

        # If we reached the destination, return the result
        if station == destination:
            return current_time, path

        # Explore neighbors
        for neighbor, travel_time in graph.get(station, []):
            new_time = current_time + travel_time
            new_path = path + [neighbor]
            heapq.heappush(heap, (new_time, neighbor, new_path))

    # If destination is unreachable
    return None, []
```

```python
time_taken, path = dijkstra_shortest_path(metro_graph, "Miyapur", "Raidurg")
print("Time:", time_taken, "minutes")
print("Path:", " -> ".join(path))
```

```python
# This assumes you have:
# - df
# - X (training features)
# - model (trained RandomForest)
# - crowd_mapping dict

crowd_mapping = {"Low": 0, "Medium": 1, "High": 2}

def predict_crowd(station_name, hour, day_type):
    sample = pd.DataFrame([{
        "station": station_name,
        "hour": hour,
        "day_type": day_type
    }])

    sample_encoded = pd.get_dummies(sample)
    sample_encoded = sample_encoded.reindex(columns=X.columns, fill_value=0)

    pred_encoded = model.predict(sample_encoded)[0]

    inv_mapping = {v: k for k, v in crowd_mapping.items()}
    return inv_mapping[pred_encoded]
```

```python
def get_route_with_crowd(graph, source, destination, hour, day_type):
    # 1. Find shortest path (time-based)
    total_time, path = dijkstra_shortest_path(graph, source, destination)
    if not path:
        return None

    # 2. Predict crowd for each station
    station_crowd = []
    crowd_to_score = {"Low": 0, "Medium": 1, "High": 2}
    total_crowd_score = 0

    for station in path:
        level = predict_crowd(station, hour, day_type)
        score = crowd_to_score[level]
        station_crowd.append((station, level))
        total_crowd_score += score

    avg_crowd_score = total_crowd_score / len(path)

    return {
        "source": source,
        "destination": destination,
        "total_time": total_time,
        "path": path,
        "station_crowd": station_crowd,
        "avg_crowd_score": avg_crowd_score
    }
```

```python
result = get_route_with_crowd(
    metro_graph,
    source="Miyapur",
    destination="Raidurg",
    hour=9,
    day_type="weekday"
)

print("Source:", result["source"])
print("Destination:", result["destination"])
print("Total Time:", result["total_time"], "minutes")
print("Route:", " -> ".join(result["path"]))
print("\nStation-wise crowd prediction:")

for st, lvl in result["station_crowd"]:
    print(f"{st}: {lvl}")

print("\nAverage crowd score (0=Low, 2=High):", result["avg_crowd_score"])
```

```python
def list_stations():
    print("Available stations:")
    for s in metro_graph.keys():
        print(" -", s)


def plan_trip_cli():
    print("=== Hyderabad Metro Crowd-Aware Route Planner ===\n")
    list_stations()
    print()

    source = input("Enter source station exactly as shown above: ").strip()
    destination = input("Enter destination station exactly as shown above: ").strip()

    if source not in metro_graph or destination not in metro_graph:
        print("\n[Error] One or both stations are invalid. Please try again.")
        return

    try:
        hour = int(input("Enter hour of travel (0-23, e.g., 9 for 9 AM, 18 for 6 PM): ").strip())
        if hour < 0 or hour > 23:
            raise ValueError
    except ValueError:
        print("\n[Error] Invalid hour. Please enter a number between 0 and 23.")
        return

    day_type = input("Enter day type (weekday/weekend): ").strip().lower()
    if day_type not in ["weekday", "weekend"]:
        print("\n[Error] Day type must be 'weekday' or 'weekend'.")
        return

    result = get_route_with_crowd(metro_graph, source, destination, hour, day_type)

    if result is None:
        print("\nNo route found between the given stations.")
        return

    print("\n=== Route Summary ===")
    print("Source:", result["source"])
    print("Destination:", result["destination"])
    print("Total Travel Time:", result["total_time"], "minutes")
    print("Route:", " -> ".join(result["path"]))

    print("\nStation-wise crowd prediction:")
    for st, lvl in result["station_crowd"]:
        print(f"  - {st}: {lvl}")

    print("\nAverage crowd score (0 = Low, 1 = Medium, 2 = High):", round(result["avg_crowd_score"], 2))


# Call this function to run the CLI planner
plan_trip_cli()
```

```python
import pandas as pd

stations_info = {
    "Miyapur": "residential",
    "JNTU College": "IT_residential",
    "KPHB Colony": "IT_residential",
    "SR Nagar": "residential",
    "Ameerpet": "interchange",
    "LB Nagar": "residential",
    "Nagole": "residential",
    "Uppal": "residential",
    "Stadium": "residential",
    "MGBS": "interchange",
    "Raidurg": "IT_hub",
    "JBS Parade Ground": "interchange",
}

day_types = ["weekday", "weekend"]

rows = []

for station, stype in stations_info.items():
    for day in day_types:
        for hour in range(6, 23):  # 6 to 22
            # Rule-based crowd label (no randomness for now)
            # Peak hours
            if day == "weekday" and (8 <= hour <= 10 or 17 <= hour <= 20):
                if stype in ["interchange", "IT_hub", "IT_residential"]:
                    crowd = "High"
                else:
                    crowd = "Medium"
            # Midday moderate hours
            elif day == "weekday" and (11 <= hour <= 16):
                if stype in ["interchange", "IT_hub", "IT_residential"]:
                    crowd = "Medium"
                else:
                    crowd = "Low"
            # Late evening / early morning
            elif day == "weekday":
                crowd = "Low"
            # Weekend pattern (generally less crowd)
            else:  # weekend
                if 10 <= hour <= 13:
                    crowd = "Medium"
                elif 18 <= hour <= 20 and stype in ["interchange", "IT_hub"]:
                    crowd = "Medium"
                else:
                    crowd = "Low"

            rows.append({
                "station": station,
                "hour": hour,
                "day_type": day,
                "crowd_level": crowd
            })

df = pd.DataFrame(rows)
print(df.head())
print("\nClass distribution:\n", df["crowd_level"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Encode target
crowd_mapping = {"Low": 0, "Medium": 1, "High": 2}
df["crowd_encoded"] = df["crowd_level"].map(crowd_mapping)

# One-hot encode features
X = pd.get_dummies(df[["station", "hour", "day_type"]])
Y = df["crowd_encoded"]

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.3, random_state=42, stratify=Y
)

model = RandomForestClassifier(
    n_estimators=300,
    max_depth=12,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, Y_train)

Y_pred = model.predict(X_test)
print("Model Accuracy:", accuracy_score(Y_test, Y_pred))
print("\nClassification Report:")
print(classification_report(Y_test, Y_pred, target_names=["Low", "Medium", "High"]))
print("\nConfusion Matrix:\n", confusion_matrix(Y_test, Y_pred))
```

```python
def predict_crowd(station_name, hour, day_type):
    sample = pd.DataFrame([{
        "station": station_name,
        "hour": hour,
        "day_type": day_type
    }])

    sample_encoded = pd.get_dummies(sample)
    sample_encoded = sample_encoded.reindex(columns=X.columns, fill_value=0)

    pred_encoded = model.predict(sample_encoded)[0]
    inv_mapping = {v: k for k, v in crowd_mapping.items()}
    return inv_mapping[pred_encoded]
```
